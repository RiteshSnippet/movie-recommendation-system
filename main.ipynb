```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess(text):
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    text = text.lower()

    words = text.split()

    words = [
        lemmatizer.lemmatize(word)
        for word in words
        if word not in stop_words
    ]

    return ' '.join(words)

df['movie_features'] = df['movie_features'].apply(preprocess)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(df['movie_features'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)
```

```python
df['normalized_title'] = df['title'].apply(
    lambda x: re.sub(r'[^a-zA-Z0-9]', '', x).lower()
)
```

```python
df.head(2)
```

```python
def get_recommendations(title, cosine_sim=cosine_sim):
    title = re.sub(r'[^a-zA-Z0-9]', '', title).lower()

    matches = df[df['normalized_title'] == title]
    
    if matches.empty:
        return "Movie not found"

    idx = matches.index[0]

    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    sim_scores = sim_scores[1:11]

    movie_indices = [i[0] for i in sim_scores]

    return df['title'].iloc[movie_indices]

print(get_recommendations('AVATAR'))

# 2405                     Aliens
# 838                      Alien³
# 3160                      Alien
# 47      Star Trek Into Darkness
# 278          Planet of the Apes
# 4336             Silent Running
# 373             Mission to Mars
# 1533                  Moonraker
# 239                     Gravity
# 541                     Soldier
# Name: title, dtype: str
```
